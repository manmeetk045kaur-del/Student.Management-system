```python
students = []  # Global list to hold every student dictionary

def add_student():
    """Prompt user for student details and add to the system."""
    try:
        sid = int(input("Enter Student ID: "))
        # Check if ID already exists
        for s in students:
            if s["id"] == sid:
                print("Error: Student ID already exists.")
                return

        name = input("Enter Name: ").strip()
        age = int(input("Enter Age: "))
        course = input("Enter Course: ").strip()
        marks_input = input("Enter Marks separated by comma (e.g., 88,92,79): ")
        marks = [int(m.strip()) for m in marks_input.split(",")]

        student = {
            "id": sid,
            "name": name,
            "age": age,
            "course": course,
            "marks": marks
        }
        students.append(student)
        print(f"Successfully added {name} (ID: {sid})")
    except ValueError:
        print("Invalid input format. Please enter correct numeric values for ID, Age, and Marks.")

def view_students():
    """Display all student records and their average marks."""
    if not students:
        print("No student records found.")
        return

    print("\n--- Student Roster ---")
    for s in students:
        avg = sum(s["marks"]) / len(s["marks"]) if s["marks"] else 0
        print(f"ID: {s['id']} | Name: {s['name']} | Age: {s['age']} | Course: {s['course']} | Marks: {s['marks']} | Avg: {avg:.1f}")
    print("-" * 25)

def update_student():
    """Update existing student details by ID."""
    try:
        sid = int(input("Enter Student ID to update: "))
        for s in students:
            if s["id"] == sid:
                print(f"Current details for {s['name']}: Age: {s['age']}, Course: {s['course']}, Marks: {s['marks']}")
                new_name = input("Enter new name (leave blank to keep current): ").strip()
                new_age = input("Enter new age (leave blank to keep current): ").strip()
                new_course = input("Enter new course (leave blank to keep current): ").strip()
                new_marks = input("Enter new marks separated by comma (leave blank to keep current): ").strip()

                if new_name:
                    s["name"] = new_name
                if new_age:
                    s["age"] = int(new_age)
                if new_course:
                    s["course"] = new_course
                if new_marks:
                    s["marks"] = [int(m.strip()) for m in new_marks.split(",")]

                print(f"Successfully updated student ID {sid}")
                return
        print("Student ID not found.")
    except ValueError:
        print("Invalid input during update.")

def delete_student():
    """Delete a student record by ID."""
    global students
    try:
        sid = int(input("Enter Student ID to delete: "))
        initial_count = len(students)
        students = [s for s in students if s["id"] != sid]
        if len(students) < initial_count:
            print(f"Successfully deleted student ID {sid}")
        else:
            print("Student ID not found.")
    except ValueError:
        print("Please enter a valid numeric ID.")

def search_student():
    """Search for students by ID or name."""
    term = input("Enter search term (ID or Name): ").strip().lower()
    results = [
        s for s in students
        if term == str(s["id"]) or term in s["name"].lower()
    ]

    if not results:
        print("No matching students found.")
        return

    print("\n--- Search Results ---")
    for s in results:
        avg = sum(s["marks"]) / len(s["marks"]) if s["marks"] else 0
        print(f"ID: {s['id']} | Name: {s['name']} | Age: {s['age']} | Course: {s['course']} | Marks: {s['marks']} | Avg: {avg:.1f}")
    print("-" * 22)

def main():
    while True:
        print("\n=== Student Management System ===")
        print("1. Add Student")
        print("2. View All Students")
        print("3. Update Student")
        print("4. Delete Student")
        print("5. Search Student")
        print("6. Exit")

        choice = input("Enter choice (1-6): ").strip()

        if choice == "1":
            add_student()
        elif choice == "2":
            view_students()
        elif choice == "3":
            update_student()
        elif choice == "4":
            delete_student()
        elif choice == "5":
            search_student()
        elif choice == "6":
            print("Exiting Student Management System. Goodbye!")
            break
        else:
            print("Invalid choice! Please enter a number between 1 and 6.")

if __name__ == "__main__":
    main()
```
